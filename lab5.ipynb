# Кореляційний аналіз

Блок коду виконує розрахунок коефіцієнтів кореляції Пірсона між віком респондента, середньомісячним доходом на члена сім'ї та особистою заробітною платою. Обчислені матриці коефіцієнтів та p-значень дозволяють зробити висновок про наявність та інтерпретацію статистично значущих зв'язків.

```python
import pyreadstat
import pandas as pd
from scipy.stats import pearsonr

df, meta = pyreadstat.read_sav('data/Kyiv1991.sav')
df.columns = df.columns.str.lower()
cols = ['v166', 'v173', 'v174']

df_clean = df[cols].dropna()

corr_matrix = df_clean.corr(method='pearson')
p_values = df_clean.corr(method=lambda x, y: pearsonr(x, y)[1])

print("Матриця кореляцій Пірсона:\n", corr_matrix)
print("\nМатриця p-значень:\n", p_values)
```
