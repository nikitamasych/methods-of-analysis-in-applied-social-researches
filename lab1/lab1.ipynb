# Завантаження даних

Імпорт необхідних бібліотек та завантаження масиву даних соціологічного дослідження з файлу.

```python
import pandas as pd
import pyreadstat
df, meta = pyreadstat.read_sav('data/Kyiv1991.SAV')
```

# Фільтрація 1: Жінки за рівнем довіри та мовою спілкування

Відбір жінок, які повністю або в основному довіряють керівництву країни та спілкуються в родині тільки українською мовою. Виведення кількості відібраних анкет.

```python
filter_1 = df[(df['V165'] == 2) & (df['V138'].isin([1, 2])) & (df['V86'] == 1)]
print(len(filter_1))
```

# Фільтрація 2: Респонденти з відповідями про прибутки

Відбір тих респондентів, що надали відповіді на обидва питання стосовно своїх прибутків без пропусків. Виведення кількості відібраних анкет.

```python
filter_2 = df[df['V173'].notna() & df['V174'].notna()]
print(len(filter_2))
```

# Фільтрація 3: Жінки з вищою освітою певного віку

Відбір жінок з вищою освітою у віковій категорії від 25 до 40 років. Виведення кількості відібраних анкет.

```python
filter_3 = df[(df['V165'] == 2) & (df['V167'] == 1) & (df['V166'] >= 25) & (df['V166'] <= 40)]
print(len(filter_3))
```
