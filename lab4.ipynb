# Аналіз зв'язку між категоріальними ознаками

Код будує таблицю спряженості для перевірки статистичного зв'язку між освітою та рівнем знайомства з принципами ринкових відносин. Здійснюється розрахунок критерію хі-квадрат, p-значення та коефіцієнта Крамера для формування змістовного висновку щодо сили та значущості цього зв'язку.

```python
import pyreadstat
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

df, meta = pyreadstat.read_sav('data/Kyiv1991.sav')
df.columns = df.columns.str.lower()

crosstab = pd.crosstab(df['v167'], df['v38'])
chi2, p, dof, expected = chi2_contingency(crosstab)

n = crosstab.sum().sum()
min_dim = min(crosstab.shape) - 1
cramer_v = np.sqrt(chi2 / (n * min_dim))

print("Таблиця спряженості:\n", crosstab)
print("\nChi-square:", chi2)
print("p-value:", p)
print("Cramer's V:", cramer_v)
```
