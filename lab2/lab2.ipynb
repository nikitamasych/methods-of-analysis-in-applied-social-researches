# Підготовка даних для аналізу майнового стану

Імпорт бібліотек, завантаження масиву даних та відбір змінних, що стосуються доходу респондентів, із вилученням порожніх значень.

```python
import pandas as pd
import pyreadstat
df, meta = pyreadstat.read_sav('data/Kyiv1991.SAV')
income_data = df[['V173', 'V174']].dropna()
```

# Описові статистики для доходу на одного члена сім'ї

Розрахунок мір центральної тенденції, варіації, показників розподілу та процентильних значень для змінної V173.

```python
stats_173 = income_data['V173'].describe(percentiles=[.25, .5, .75])
stats_173.loc['mode'] = income_data['V173'].mode()[0]
stats_173.loc['variance'] = income_data['V173'].var()
stats_173.loc['skewness'] = income_data['V173'].skew()
stats_173.loc['kurtosis'] = income_data['V173'].kurt()
print(stats_173)
```

# Описові статистики для особистої заробітної плати

Розрахунок мір центральної тенденції, варіації, показників розподілу та процентильних значень для змінної V174.

```python
stats_174 = income_data['V174'].describe(percentiles=[.25, .5, .75])
stats_174.loc['mode'] = income_data['V174'].mode()[0]
stats_174.loc['variance'] = income_data['V174'].var()
stats_174.loc['skewness'] = income_data['V174'].skew()
stats_174.loc['kurtosis'] = income_data['V174'].kurt()
print(stats_174)
```
