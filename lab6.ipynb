# Регресійний аналіз

Код генерує модель множинної лінійної регресії, де залежною змінною виступає рівень задоволеності життям, а незалежними є фактори самовизначення та матеріального стану. Виводяться повні таблиці результатів для нестандартизованого та стандартизованого рівнянь із зазначенням коефіцієнта детермінації для виявлення найбільш впливових змінних.

```python
import pyreadstat
import pandas as pd
import statsmodels.api as sm
from sklearn.preprocessing import StandardScaler

df, meta = pyreadstat.read_sav('data/Kyiv1991.sav')
df.columns = df.columns.str.lower()
cols = ['v21', 'v4', 'v5', 'v7', 'v8', 'v28']

df_clean = df[cols].dropna()

y = df_clean['v21']
X = df_clean[['v4', 'v5', 'v7', 'v8', 'v28']]
X_with_const = sm.add_constant(X)

model_unstandardized = sm.OLS(y, X_with_const).fit()
print("--- Нестандартизована модель ---\n")
print(model_unstandardized.summary())

scaler = StandardScaler()
df_std = pd.DataFrame(scaler.fit_transform(df_clean), columns=cols)

y_std = df_std['v21']
X_std = df_std[['v4', 'v5', 'v7', 'v8', 'v28']]

model_standardized = sm.OLS(y_std, X_std).fit()
print("\n--- Стандартизована модель ---\n")
print(model_standardized.summary())
```
