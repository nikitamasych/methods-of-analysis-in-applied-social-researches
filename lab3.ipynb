# Аналіз одновимірних розподілів

Цей код завантажує набір даних та обчислює частотні розподіли з відсотковими значеннями для вказаних змінних. Результати дозволяють описати ставлення респондентів до різних шляхів подолання економічної кризи.

```python
import pyreadstat

df, meta = pyreadstat.read_sav('data/Kyiv1991.sav')
df.columns = df.columns.str.lower()
cols = [f'v{i}' for i in range(40, 51)]

for col in cols:
    freq = df[col].value_counts(normalize=True) * 100
    print(f"--- {col} ---")
    print(freq, "\n")
```
